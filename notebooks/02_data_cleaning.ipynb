# Data Cleaning

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/cardio_train.csv", sep=";")
```

```python
df.shape
```

```python
cleaned_df = df.copy()
```

## Clean invalid blood-pressure records

```python
bp_invalid = (
    (cleaned_df["ap_hi"] <= 0) |
    (cleaned_df["ap_lo"] <= 0) |
    (cleaned_df["ap_hi"] < cleaned_df["ap_lo"]) |
    (~cleaned_df["ap_hi"].between(70, 250)) |
    (~cleaned_df["ap_lo"].between(40, 150))
)
```

```python
print("Records with invalid BP:", bp_invalid.sum())
print("Percentage:", f"{bp_invalid.mean() * 100:.2f}%")
```

```python
original_shape = cleaned_df.shape

original_shape
```

```python
cleaned_df = cleaned_df[~bp_invalid].copy()
```

```python
cleaned_df.shape
```

## Cleaning Verification

```python
# Check for remaining non-positive BP values:

print("Non-positive systolic BP:", (cleaned_df["ap_hi"] <= 0).sum())
print("Non-positive diastolic BP:", (cleaned_df["ap_lo"] <= 0).sum())
```

```python
# Check systolic/diastolic relationship:

print("Systolic BP < diastolic BP:",
      (cleaned_df["ap_hi"] < cleaned_df["ap_lo"]).sum())
```

```python
# Check the defined BP ranges:

print("Systolic BP outside 70–250:",
      (~cleaned_df["ap_hi"].between(70, 250)).sum())

print("Diastolic BP outside 40–150:",
      (~cleaned_df["ap_lo"].between(40, 150)).sum())
```

```python
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Duplicate IDs:", cleaned_df["id"].duplicated().sum())
```

## Before and After Cleaning

```python
print("Original shape:", original_shape)
print("Cleaned shape:", cleaned_df.shape)
print("Records removed:", original_shape[0] - cleaned_df.shape[0])
print(
    "Percentage removed:",
    f"{(original_shape[0] - cleaned_df.shape[0]) / original_shape[0] * 100:.2f}%"
)
```

## Save Cleaned Dataset

```python
cleaned_df.to_csv("../data/processed/cardio_cleaned.csv", index=False)
```
